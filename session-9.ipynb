```python
import os
import pandas as pd
import numpy as np

np.random.seed(42)

if not os.path.exists('gym_members_v2.csv'):
    fallback_df = pd.DataFrame({
        'member_id': range(1, 121),
        'plan': np.random.choice(['Basic', 'Standard', 'Premium'], size=120),
        'name_clean': np.random.choice(
            ['Riya Shah', 'Karan Patel', 'Neha Joshi', 'Arjun Mehta', 'Priya Desai'],
            size=120
        ),
        'goal_clean': np.random.choice(['weight loss', 'muscle gain', 'stamina'], size=120)
    })
    fallback_df[['first_name', 'last_name']] = fallback_df['name_clean'].str.split(' ', expand=True)
    fallback_df.to_csv('gym_members_v2.csv', index=False)

gym_df = pd.read_csv('gym_members_v2.csv')

days = np.random.randint(1, 29, size=len(gym_df))
months = np.random.randint(1, 13, size=len(gym_df))
years = np.random.choice([2024, 2025], size=len(gym_df))

join_date_str = [f"{d:02d}/{m:02d}/{y}" for d, m, y in zip(days, months, years)]
gym_df['join_date'] = pd.to_datetime(join_date_str, format='%d/%m/%Y')

print(gym_df.dtypes)
```

```python
gym_df['join_year'] = gym_df['join_date'].dt.year
gym_df['join_month'] = gym_df['join_date'].dt.month
gym_df['join_weekday'] = gym_df['join_date'].dt.day_name()

print(gym_df['join_year'].value_counts())
```

```python
month_starts = pd.date_range(start='2025-01-01', periods=12, freq='MS')

h1_joiners = gym_df[
    (gym_df['join_date'] >= month_starts[0]) &
    (gym_df['join_date'] < month_starts[6])
]

print(len(h1_joiners))
```

```python
as_of = pd.Timestamp('2026-01-01')

gym_df['tenure_days'] = (as_of - gym_df['join_date']).dt.days
gym_df['tenure_months'] = gym_df['tenure_days'] // 30

print(gym_df['tenure_days'].min(), gym_df['tenure_days'].max())

gym_df.to_csv('gym_members_v3.csv', index=False)
```
